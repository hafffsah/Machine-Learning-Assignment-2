```python
import pandas as pd # type: ignore
import numpy as np
'''
import sys
!{sys.executable} -m pip install scikit-learn
'''
from sklearn import model_selection
from sklearn import preprocessing

data = pd.read_csv("../ML - Assignment 2/Churn_Modelling.csv")  # cleaned data
print("Dataset size- ", data.shape)
```

```python
y = data.iloc[:, -1]
x = data.iloc[:, 3:-1]
```

```python
encoder = preprocessing.OneHotEncoder(sparse_output= np.False_)

string_type_data = x.select_dtypes(include= ['str']).columns
encoded_data = encoder.fit_transform(x[string_type_data])

encoded_df = pd.DataFrame(
    encoded_data,
    columns= encoder.get_feature_names_out(string_type_data)
)

final_features = pd.concat(
    [x.drop(columns= string_type_data), encoded_df],
    axis=1
)

'''
2 Categorical features-
3 unique values - Geography (France, Spain, Germany)
2 - Gender
'''
```

```python
x_train, x_test, y_train, y_test = model_selection.train_test_split(final_features, y, test_size= 0.2)   

scaler = preprocessing.StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
'''
LOGISTIC REGRESSION MODEL
'''
from sklearn import linear_model
log_regression = linear_model.LogisticRegression()
log_regression.fit(x_train_scaled, y_train)

LR_prediction = log_regression.predict(x_test_scaled)


'''
DECISION TREE
'''
from sklearn import tree
decision_tree = tree.DecisionTreeClassifier()
decision_tree.fit(x_train_scaled, y_train)

DT_prediction = decision_tree.predict(x_test_scaled)


'''
K-NN CLASSIFIER
'''
from sklearn import neighbors
k_nn = neighbors.KNeighborsClassifier()
k_nn.fit(x_train_scaled, y_train)

KNN_prediction = k_nn.predict(x_test_scaled)


'''
NAIVE BAYES
'''
from sklearn import naive_bayes
NB = naive_bayes.GaussianNB()   # not multinomial (for frequency)
NB.fit(x_train_scaled, y_train)

NB_prediction = NB.predict(x_test_scaled)


'''
SUPPORT VECTOR MACHINE
'''
from sklearn import svm
SVM = svm.SVC()
SVM.fit(x_train_scaled, y_train)
SVM_prediction = SVM.predict(x_test_scaled)
```

```python
# ----- Evaluation Metrics ------
from sklearn import metrics

predictions = [LR_prediction, DT_prediction, KNN_prediction, NB_prediction, SVM_prediction]
eval_metrics = {
    "accuracy_score": [],
    "auc_score": [],
    "recall_score": [],
    "precision_score": [],
    "f1_score": [],
    "mcc": []
}

for model in predictions:
    eval_metrics["accuracy_score"].append(metrics.accuracy_score(y_test , model))
    eval_metrics["auc_score"].append(metrics.roc_auc_score(y_test , model))
    eval_metrics["recall_score"].append(metrics.recall_score(y_test , model))
    eval_metrics["precision_score"].append(metrics.precision_score(y_test , model))
    eval_metrics["f1_score"].append(metrics.f1_score(y_test , model))
    eval_metrics["mcc"].append(metrics.matthews_corrcoef(y_test , model))
```
